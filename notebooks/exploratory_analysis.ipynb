# Settlement Fail Analytics — Exploratory Analysis

Exploratory notebook on top of the synthetic trade-settlement dataset produced by `pipeline.py`. See the project README for dataset methodology and assumptions.


```python
import sys
sys.path.append('..')

import pandas as pd
import matplotlib.pyplot as plt

from pipeline import generate_trades, compute_analytics

df = generate_trades()
analytics = compute_analytics(df)
df.head()
```

## Headline numbers

```python
print(f"Total trades: {analytics['total_trades']:,}")
print(f"Total fails: {analytics['total_fails']:,}")
print(f"Overall fail rate: {analytics['overall_fail_rate']:.2%}")
print(f"Estimated cost of fails: ${analytics['total_estimated_cost']:,.2f}")
```

## Fail rate by asset class

```python
analytics['fail_rate_by_asset_class'].plot(kind='barh', title='Fail Rate by Asset Class')
plt.xlabel('Fail Rate')
plt.gca().xaxis.set_major_formatter(plt.matplotlib.ticker.PercentFormatter(1.0))
plt.tight_layout()
plt.show()
```

## Weekly fail rate trend

```python
analytics['fail_rate_trend'].plot(marker='o', title='Weekly Fail Rate Trend')
plt.ylabel('Fail Rate')
plt.gca().yaxis.set_major_formatter(plt.matplotlib.ticker.PercentFormatter(1.0))
plt.tight_layout()
plt.show()
```

## Aging distribution

```python
analytics['aging_distribution'].plot(kind='bar', title='Aging Distribution of Fails')
plt.ylabel('Fail Count')
plt.tight_layout()
plt.show()
```

## Root cause breakdown

```python
analytics['root_cause_breakdown'].plot(kind='pie', autopct='%1.1f%%', ylabel='')
plt.title('Root Cause Breakdown')
plt.tight_layout()
plt.show()
```
